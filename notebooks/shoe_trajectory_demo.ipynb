## Cell 2: Quick Test

```python
import sys
sys.path.insert(0, '.')
import numpy as np
from src.pointcloud_generator import generate_sole_shape
from src.multi_plane_projection import multi_plane_projection
from src.trajectory_extraction import extract_trajectory, compute_trajectory_stats, format_trajectory_stats
from src.visualization import visualize_full_pipeline

full_cloud, edge_cloud = generate_sole_shape('sneaker')
print(f'Point cloud: {full_cloud.shape[0]} points')

proj = multi_plane_projection(edge_cloud, num_planes=8)
contour = proj['contour']
print(f'Contour: {len(contour)} points')

traj = extract_trajectory(contour, min_points=50, max_points=150)
stats = compute_trajectory_stats(traj['trajectory'])
print(format_trajectory_stats(stats))

img = visualize_full_pipeline(full_cloud, contour, traj['trajectory'], traj['normals'])
from IPython.display import Image, display
display(Image(img))
print('Test passed')
```
